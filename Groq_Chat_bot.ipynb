# Import the Packages

```python
import os, getpass

def load_groq_key() -> str:
    """Find the Groq key without ever printing it."""
    if os.getenv("GROQ_API_KEY"):                       # 1. already in the environment
        return os.environ["GROQ_API_KEY"]
    try:                                                # 2. Google Colab "Secrets" panel
        from google.colab import userdata
        key = userdata.get("GROQ_API_KEY")
        if key:
            return key
    except Exception:
        pass
    try:                                                # 3. a local .env file
        from dotenv import load_dotenv
        load_dotenv()
        if os.getenv("GROQ_API_KEY"):
            return os.environ["GROQ_API_KEY"]
    except ImportError:
        pass
    return getpass.getpass("Paste your Groq API key (input is hidden): ")   # 4. ask

os.environ["GROQ_API_KEY"] = load_groq_key().strip()
key = os.environ["GROQ_API_KEY"]
print("Key loaded:", "yes" if key else "NO", "| starts with gsk_:", key.startswith("gsk_"), "| length:", len(key))
```

```python
import gradio
from groq import Groq
```

```python
client = Groq()
```

#Define a function to give content and role

```python
def initialize_messages():
    return [{"role": "system",
             "content": """You are a skilled criminal lawyer with a
             successful track record in numerous cases. Your role is to
             assist people by providing guidance on Indian laws and
             offering answers in a professional legal manner."""}]
```

#Assign it to a variable

```python
messages_prmt = initialize_messages()
```

#Define a function to connect with LLM

```python
def customLLMBot(user_input, history):
    global messages_prmt

    messages_prmt.append({"role": "user", "content": user_input})

    response = client.chat.completions.create(
        messages=messages_prmt,
        model="openai/gpt-oss-120b",
    )
    print(response)
    LLM_reply = response.choices[0].message.content
    messages_prmt.append({"role": "assistant", "content": LLM_reply})

    return LLM_reply
```

#Create an object of chat interface class in gradio

```python
iface = gradio.ChatInterface(customLLMBot,
                     chatbot=gradio.Chatbot(height=300),
                     textbox=gradio.Textbox(placeholder="Ask me a question related to law"),
                     title="Lawyer ChatBot",
                     description="Chat bot for law assistance",
                     examples=["hi","What is IPC sessions", "how to get a bail"]
                     )
```

#Call launch function to execute

```python
iface.launch(share=True)
```
